Olá! Segue exercício da atividade Python.

A situação escolhida para executar as exigências, foi o cenário de término de campeonato brasileiro, onde o time a ser inserido terá 35 pontos, está lutando contra o rebaixamento e os cálculos feitos irão exibir a % de aproveitamento, pontos conseguido e o resultado se o time escapou ou não do rebaixamento.

```python
# Dados fixos do cenário
PONTOS_ATUAIS = 35
JOGOS_RESTANTES = 5
PONTOS_SEGURANCA = 45  # meta aproximada para escapar do rebaixamento

time_do_coracao = input('Olá, Tudo bem? Insira aqui o nome do seu time: ')

continuar = 's'

while continuar.lower() == 's':
    # Entradas numéricas convertidas para int
    pontos_recentes = int(input('Quantos pontos o time fez nas últimas 5 partidas? '))
    vitorias = int(input('Quantas vitórias o time teve nessas partidas? '))

    # Operações matemáticas
    aproveitamento = (vitorias / JOGOS_RESTANTES) * 100
    media_pontos = pontos_recentes / JOGOS_RESTANTES                 # divisão
    pontos_projetados = media_pontos * JOGOS_RESTANTES               # multiplicação
    total_projetado = PONTOS_ATUAIS + pontos_projetados              # soma
    pontos_faltando = PONTOS_SEGURANCA - PONTOS_ATUAIS               # subtração
    pontos_em_disputa = JOGOS_RESTANTES * 3
    aproveitamento_necessario = (pontos_faltando / pontos_em_disputa) * 100

    # Saída
    print(f'\n--- Análise do {time_do_coracao} ---')
    print(f'A porcentagem de aproveitamento do seu time seria de: {aproveitamento}% ')
    print(f'Média de pontos nas últimas {JOGOS_RESTANTES} partidas: {media_pontos:.2f}')
    print(f'Número de vitórias nas últimas {JOGOS_RESTANTES} partidas: {vitorias}')
    print(f'Aproveitamento necessário nos {JOGOS_RESTANTES} jogos que faltavam para evitar rebaixamento: {aproveitamento_necessario:.1f}%')
    print(f'Pontuação projetada ao fim do campeonato: {total_projetado:.1f}')

    # Condicional para avaliar o rebaixamento corrigido
    if total_projetado >= PONTOS_SEGURANCA:
        print('Situação: Mantendo esse ritmo, seu time deve escapar do rebaixamento, fique esperançoso!')
    elif time_do_coracao.lower() == 'vasco da gama' and total_projetado < PONTOS_SEGURANCA:
        print('Porquê deus odeia o Vaaascooo?! Infelizmente fomos rebaixados meu caro.')
    else:
        print('Situação: Eeeita, preocupante hein! Nesse ritmo, o time continua na zona de rebaixamento!')

    continuar = input('\nDeseja fazer outra análise? (s/n): ')

print('Maravilha! Foi um prazer auxiliá-lo hoje. Boa sorte ao seu time!')
```
